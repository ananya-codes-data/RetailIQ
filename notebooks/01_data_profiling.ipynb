# Setup

Importing libraries

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)
```

# 2. Table viewing and basic info
Viewing the first 5 rows of the table along with count total of rows and columns

```python
df1 = pd.read_excel("../data/raw/online_retail_II.xlsx", sheet_name="Year 2009-2010")
print(f"year_2009_2010: {df1.shape[0]} rows, {df1.shape[1]} columns")
display(df1.head(5))
```

```python
df2 = pd.read_excel("../data/raw/online_retail_II.xlsx", sheet_name="Year 2010-2011")
print(f"year_2010_2011: {df2.shape[0]} rows, {df2.shape[1]} columns")
display(df2.head(5))
```

```python
# Inspecting the data types and null values
print(f"table: retail_2009_2010")
df1.info()
```

```python
# Inspecting the data types and null values
print(f"table: retail_2010_2011")
df2.info()
```

# 3. Appending both the tables

```python
retail_transactions = pd.concat([df1, df2], ignore_index=True)
print(f"retail_transactions: {retail_transactions.shape[0]} rows, {retail_transactions.shape[1]} columns")
display(retail_transactions.head(5))
```

# 4. Missing Audit

```python
# Inspecting null values in each column of the retail_transactions DataFrame
print(f"table: retail_transactions")
print(retail_transactions.isnull().sum())
```

```python
# Calculating the percentage of null values in each column of the retail_transactions DataFrame
print(f"table: retail_transactions")
null_values = (retail_transactions.isnull().mean() * 100).round(2)
print(null_values[null_values > 0].sort_values(ascending=False))
```

```python
# Identifying rows with null values in both "Description" and "Customer ID" columns
retail_transactions[retail_transactions["Description"].isnull() & retail_transactions["Customer ID"].isnull()].reset_index(drop=True)
```

# 5. Duplicate checks

```python
print(f"table: retail_transactions")
print(f"duplicates: {retail_transactions.duplicated().sum()}")  # Count of duplicate rows
print(f"{retail_transactions.duplicated().mean() * 100:.2f}% of rows are duplicates")   # Percentage of duplicate rows
```

# 6. Categorical value inspection
Checking the distribution of categorical variables

```python
retail_transactions["Country"].value_counts(dropna=False)
```

From the categorical value inspection it is found that, we are getting maximum customers from UK, so the data would naturally be more inclined towards UK.

# 8. Profiling of few columns

```python
# Identifying rows with "Invoice" starting with "C" as they represent canceled transactions
retail_transactions[retail_transactions["Invoice"].str.startswith("C", na=False)].reset_index(drop=True)
```

```python
# Identifying rows with "Invoice" starting with "C" and null "Customer ID"
retail_transactions[(retail_transactions["Invoice"].str.startswith("C", na=False)) & 
                    (retail_transactions["Customer ID"].isnull())].reset_index(drop=True)
```

```python
# Identifying rows with "Invoice" not starting with "C" and negative "Quantity" values
retail_transactions[~(retail_transactions["Invoice"].str.startswith("C", na=False)) & 
                    (retail_transactions["Quantity"] < 0)].reset_index(drop=True)
```

```python
# Identifying rows with "Invoice" not starting with "C" and negative "Quantity" values and null "Customer ID"
retail_transactions[~(retail_transactions["Invoice"].str.startswith("C", na=False)) & 
                    (retail_transactions["Quantity"] < 0) & 
                    (retail_transactions["Customer ID"].isnull())].reset_index(drop=True)
```
