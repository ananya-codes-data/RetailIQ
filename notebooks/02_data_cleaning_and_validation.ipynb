# Phase 2 — Data cleaning

# Set up libraries

```python
import pandas as pd
import numpy as np

# Set pandas display options for better visibility of DataFrames
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)
```

# Load datasets from Excel file

```python
df1 = pd.read_excel("../data/raw/online_retail_II.xlsx", sheet_name="Year 2009-2010")
print(f"year_2009_2010: {df1.shape[0]} rows, {df1.shape[1]} columns")
display(df1.head(5))
```

```python
df2 = pd.read_excel("../data/raw/online_retail_II.xlsx", sheet_name="Year 2010-2011")
print(f"year_2010_2011: {df2.shape[0]} rows, {df2.shape[1]} columns")
display(df2.head(5))
```

# Appending both the tables

```python
retail_transactions = pd.concat([df1, df2], ignore_index=True)
print(f"retail_transactions: {retail_transactions.shape[0]} rows, {retail_transactions.shape[1]} columns")
display(retail_transactions.head(5))
```

# Creating a copy of the raw table

```python
df = retail_transactions.copy(deep=True)
```

```python
# Display the DataFrame
df
```

```python
# Display the DataFrame information
df.info()
```

# 1. Header Cleaning

```python
print(df.columns.to_list())

df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

print("FIX APPLIED: Column names have been standardized to lowercase with underscores instead of spaces.")
print(df.columns.to_list())
```

# 2. Standardize data types

```python
df["invoice"] = df["invoice"].astype(str)

df["stockcode"] = df["stockcode"].astype(str)

df["description"] = df["description"].astype(str)

df["customer_id"] = df["customer_id"].astype("Int64")

print("Data types changed")
df.info()
```

# 3. Removing whitespace

```python
df["description"] = df["description"].str.strip()
```

# 4. Adding flags

```python
df["is_cancellation"] = df["invoice"].str.startswith("C", na=False)
```

```python
# StockCodes in Online Retail II that represent non-product line items
NON_PRODUCT_CODES = {
    "POST", "DOT", "M", "m", "D", "C2", "BANK CHARGES",
    "AMAZONFEE", "CRUK", "PADS", "S", "B", "ADJUST", "ADJUST2",
}
```

```python
df["is_non_product"] = df["stockcode"].isin(NON_PRODUCT_CODES)
```

```python
df["is_missing_customer"] = df["customer_id"].isna()
```

```python
df["is_customer_cancellation"] = df["is_cancellation"] & df["customer_id"].notna()
```

# 5. Dropping duplicates, impossible columns, etc

```python
# Drop exact duplicate rows, keeping the last occurrence
before = len(df)
df = df.drop_duplicates(keep='last').reset_index(drop=True)
print(f"Dropped {before - len(df):,} exact duplicate rows")
```

```python
# Drop rows with zero information: no Description, no Customer ID, Price == 0

zero_info_mask = (df["description"].isna() & df["customer_id"].isna() & (df["price"] == 0))
print(f"Dropping {zero_info_mask.sum():,} zero-information rows "
          f"(no Description, no Customer ID, Price == 0)")
df = df[~zero_info_mask].reset_index(drop=True)
```

```python
# Stock adjustment descriptions that are known to be junk or non-informative

ADJUSTMENT_DESCRIPTIONS = {
    "lost", "missing", "smashed", "check", "short", "damages",
    "damaged", "found", "thrown away", "wet", "crushed",
}
```

```python
# stock adjustment: NOT a cancellation, negative qty, price == 0,
# and either missing description or a known junk description

junk_desc_mask = df["description"].str.lower().isin(ADJUSTMENT_DESCRIPTIONS)
df["is_stock_adjustment"] = (
        (~df["is_cancellation"])
        & (df["quantity"] < 0)
        & (df["price"] == 0)
        & (df["description"].isna() | junk_desc_mask))
```

```python
# Display the cleaned and validated DataFrame
df
```

# Exporting the cleaned dataset to the interim folder

```python
df.to_csv("../data/interim/cleaned_retail_transactions.csv", index=False)
```
