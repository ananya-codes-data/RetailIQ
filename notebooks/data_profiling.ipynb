# 3. Missing Audit

```python
print(f"table: year_2009_2010")
print(year_2009_2010.isnull().sum())
```

```python
print(f"table: year_2009_2010")
null_values_y1 = year_2009_2010.isnull().mean() * 100
print(null_values_y1[null_values_y1 > 0].sort_values(ascending=False))
```

```python
print(f"table: year_2010_2011")
print(year_2010_2011.isnull().sum())
```

```python
print(f"table: year_2010_2011")
null_values_y2 = year_2010_2011.isnull().mean() * 100
print(null_values_y2[null_values_y2 > 0].sort_values(ascending=False))
```

# 4. Duplicate checks

```python
print(f"table: year_2009_2010")
print(year_2009_2010.duplicated().sum())
print(f"{year_2009_2010.duplicated().mean() * 100:.2f}% of rows are duplicates")

print(f"table: year_2010_2011")
print(year_2010_2011.duplicated().sum())
print(f"{year_2010_2011.duplicated().mean() * 100:.2f}% of rows are duplicates")
```
